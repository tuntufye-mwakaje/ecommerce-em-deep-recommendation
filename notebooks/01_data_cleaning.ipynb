# 01 - Data Cleaning

## E-Commerce Personalization Using EM Clustering and Deep Recommendation

### Objective

This notebook performs the initial cleaning and preprocessing of the
Online Retail transaction dataset used in the E-Commerce Personalization
project.

### Tasks

- Load the dataset
- Inspect the dataset structure
- Identify missing values
- Remove duplicate records
- Handle invalid transactions
- Prepare the cleaned dataset for further analysis

```python
# Step 1 - Data Preparation
# Required Libraries
import pandas as pd
import numpy as np

# Load the dataset

file_path = "OnlineRetail.csv"
data = pd.read_csv(file_path, encoding='ISO-8859-1')

# Display dataset shape
print("Dataset Shape:", data.shape)
```

```python
# Check for missing values
print("Missing values in each column:\n", data.isnull().sum())
```

```python
# Step 2 - Drop rows with missing CustomerID (as they can’t be used for user clustering)

data = data.dropna(subset=['CustomerID'])

# Remove duplicate rows (if any)
data = data.drop_duplicates()

# Remove negative or zero Quantity and UnitPrice values (returns or data errors)
data = data[(data['Quantity'] > 0) & (data['UnitPrice'] > 0)]

# Confirm the cleaning process
print("✅ Data cleaned successfully!")
print("Remaining rows:", data.shape[0])
print("Remaining columns:", data.shape[1])

# Check if any invalid values remain
print("\nNumber of rows with Quantity <= 0:", (data['Quantity'] <= 0).sum())
print("Number of rows with UnitPrice <= 0:", (data['UnitPrice'] <= 0).sum())
```

```python
# Step 3 — Create new features for customer-level analysis

# Add a new feature: TotalPrice = Quantity × UnitPrice
data['TotalPrice'] = data['Quantity'] * data['UnitPrice']

# Group data by CustomerID
customer_df = data.groupby('CustomerID').agg({
    'InvoiceNo': 'nunique',       # Number of unique purchases
    'Quantity': 'sum',            # Total quantity purchased
    'TotalPrice': 'sum',          # Total money spent
    'UnitPrice': 'mean'           # Average unit price paid
}).reset_index()

# Rename columns for readability
customer_df.rename(columns={
    'InvoiceNo': 'NumPurchases',
    'Quantity': 'TotalQuantity',
    'TotalPrice': 'TotalSpending',
    'UnitPrice': 'AvgUnitPrice'
}, inplace=True)

# Display first few records
print("\n📊 Customer-level aggregated dataset:")
display(customer_df.head())
```

```python
# Step 4 - Save Cleaned data 
customer_df.to_csv("cleaned_data.csv", index=False)
```

## Cleaning Summary

The dataset was cleaned and prepared for the next stage of the
E-Commerce Personalization workflow.
